```python
from statsmodels.stats.power import TTestPower
import numpy as np

# Set your experimental parameters
effect_size = 0.74  # Cohen's d (e.g., 0.50 for a conservative estimate, or 0.74 based on original findings)
alpha = 0.05        # Significance level (two-tailed)
power = 0.95        # Desired statistical power (90%)

# Initialize power analysis tool for paired/dependent means
power_analysis = TTestPower()
  
# Calculate required sample size
sample_size = power_analysis.solve_power(
    effect_size=effect_size, 
    power=power, 
    alpha=alpha, 
    # alternative='two-sided'
)

print(f"Required sample size (N): {int(np.ceil(sample_size))}")
```

```python
import math
from statsmodels.stats.power import TTestPower, TTestIndPower

paired_power = TTestPower()
ind_power = TTestIndPower()

def calc_paired_n(d, alpha=0.05, power=0.95, alternative='larger'):
    """Calculates N for a paired-samples t-test (within-subjects)."""
    n = paired_power.solve_power(effect_size=d, alpha=alpha, power=power, alternative=alternative)
    return math.ceil(n)

def calc_ind_n_per_group(d, alpha=0.05, power=0.95, alternative='two-sided'):
    """Calculates N per group for an independent-samples t-test / contrast."""
    n = ind_power.solve_power(effect_size=d, alpha=alpha, power=power, alternative=alternative)
    return math.ceil(n)

print("=== AUTHORS' A PRIORI POWER ANALYSES ===")

# Experiment 1: Two separate outcome conditions (Statement Truth vs. Source Credibility)
# Each condition runs a one-tailed paired-samples t-test (d = 0.35, alpha = .05, power = .95)
n_exp1_per_cond = calc_paired_n(d=0.35, alpha=0.05, power=0.95, alternative='larger')
print(f"Exp 1 (A Priori): {n_exp1_per_cond} per condition -> Total N = {n_exp1_per_cond * 2}")

# Experiment 2: Single within-subjects design testing downstream source credibility
# One-tailed paired-samples t-test (d = 0.30, alpha = .05, power = .95)
n_exp2 = calc_paired_n(d=0.30, alpha=0.05, power=0.95, alternative='larger')
print(f"Exp 2 (A Priori): Total N = {n_exp2} (Authors oversampled to 130)")

# Experiment 3: Single within-subjects design (passive exposure, no initial truth ratings)
# One-tailed paired-samples t-test (d = 0.25, alpha = .05, power = .95)
n_exp3 = calc_paired_n(d=0.25, alpha=0.05, power=0.95, alternative='larger')
print(f"Exp 3 (A Priori): Total N = {n_exp3} (Authors oversampled to 180)")

# Experiment 4: 3-group between-subjects design comparing difference scores (Repeated - Unrepeated)
# Powered on contrasts for d = 0.30, alpha = .05, nominal power = .95 (to achieve .90 joint power across 2 contrasts)
# Note: In R's `pwr::pwr.t.test(d=0.30, sig.level=0.05, power=0.95, type="two.sample", alternative="greater")`:
n_exp4_per_group = calc_ind_n_per_group(d=0.30, alpha=0.05, power=0.95, alternative='larger')
print(f"Exp 4 (A Priori): {n_exp4_per_group} per group -> Total N = {n_exp4_per_group * 3}")


print("\n=== CONSERVATIVE REPLICATION POWERED FOR OBSERVED EFFECT SIZES ===")
# In the General Discussion, the authors note that the downstream credibility effect is actually small
# (e.g., d = 0.12 overall in Exp 4, and d = 0.08 to 0.15 within individual conditions).

# Exp 1 Replication (using observed direct credibility effect d = 0.74, or conservative d = 0.50)
n_rep1 = calc_paired_n(d=0.74, alpha=0.05, power=0.95, alternative='larger')
print(f"Exp 1 Replication (for observed d = 0.74): {n_rep1} per condition -> Total N = {n_rep1 * 2}")

# Exp 2 Replication (using observed downstream credibility effect d = 0.32)
n_rep2 = calc_paired_n(d=0.32, alpha=0.05, power=0.95, alternative='larger')
print(f"Exp 2 Replication (for observed d = 0.32): Total N = {n_rep2}")

# Exp 3 / Exp 4 Simple Effect Replication (using observed downstream d = 0.14 in Paired-Only)
n_rep3 = calc_paired_n(d=0.14, alpha=0.05, power=0.95, alternative='larger')
print(f"Exp 3 / Single-Condition Replication (for observed d = 0.14): Total N = {n_rep3}")

# Exp 4 Overall Main Effect Replication (using observed main effect d = 0.12 across 3 groups)
n_rep4_total = calc_paired_n(d=0.12, alpha=0.05, power=0.95, alternative='larger')
n_rep4_per_group = math.ceil(n_rep4_total / 3)
print(f"Exp 4 Main Effect Replication (for observed d = 0.12): ~{n_rep4_per_group} per group -> Total N = {n_rep4_per_group * 3}")
```
